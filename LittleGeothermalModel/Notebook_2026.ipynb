## 2. Definitions
In the next cell, we define our model.
For example:
- How deep do we want to drill?
- How thick is the geological layer?
- How far should the wells be spaced?

These things are called _parameters_.

The most important parameters are listed here and you can type in the cell to change the numbers. There are more parameters to change if you're curious. See below.

```python
# Code cell 2 - Definitions

# Reservoir depth: The depth of the top of the reservoir in m.
reservoir_depth = 2300

# Reservoir thickness in m.
reservoir_thick = 80

# porosity in percent
porosity=20

# Permeability in mDarcy
permeability = 600

# well spacing. The distance between injector and producer in m
wells_x = 800

# injection temperature. the temperature (in deg Celsius) of the cold water that is being re-injected.
injT = 32

# flow rate. How much water (in m3/h) we pump through the reservoir
flow=200

print("\n\nDefinitions complete!")
```

## 3 - Make the doublet
In this cell, we take all the parameters and give them to the doublet function. After executing, "doublet 1" is an object that has all the parameters we defined above.

**Important:** If you change parameters, you have to run this cell again to see the changes.

```python
# Code cell 3 - Make the doublet object
# Give all this input to the function:
my_doub = Doublet(depth_m=reservoir_depth, thickn_m=reservoir_thick, poro_percent=porosity, perm_mD=permeability, w_space=wells_x, flow_m3_h=flow, T_inj=injT)



print('Doublet created')
```

## 4 - Physics
Now we will look at the physics of the doublet. We want to know
1. Production temperature.
2. Lifetime of the doublet. (Lifetime means the number of years for which the production well gives us hot water.)
3. The power (in megawatt, MW) that is needed to run the system. (Most of the time, we talk about how much power (=energy per time) a geothermal doublet produces. But it also costs power to run the pumps that move the water. The difference, "power_out - power_in" is how much we actually get.)

```python
# Production temperature
Temp_prod = my_doub.T_prod
print('T_out: %s °C' % round(Temp_prod, 2))
```

```python
# time of thermal breakthrough
time_breakth = my_doub.t_breakthrough()
print('Time of thermal breakthrough: %s years' % round(time_breakth, 2))
```

```python
# Power

# 1) How much power do we generate?
p_out = my_doub.p_doublet()
print('Power produced by the doublet: %s MW'  %round(p_out, 2))

# 2) how much power is needed to run the pumps?
p_in = my_doub.p_pumps()
print('Power consumed by the doublet: %s MW' %round(p_in, 2))

# 3) The difference:
p_net = p_out - p_in
print('Net power produced: %s MW' %round(p_net, 2))
```
