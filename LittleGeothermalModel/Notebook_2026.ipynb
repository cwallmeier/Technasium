## 2. Definitions
In the next cell, we define the details of the model.
For example:
- How deep underground is the target layer?
- How thick is the geological layer?
- How far should the wells be apart?

These things are called _parameters_.

The most important parameters are listed  here. You can type in the cell to change the numbers. There are more parameters to change if you're curious. See below.

```python
# Code cell 2 - Definitions

# Depth: The depth (of the top) of the target layer, in meters.
depth = 2300

# The thickness of the target layer, in meters.
thickn = 80

# The average porosity of the target layer, in percent.
porosity=18

# The average permeability of the target layer, in mDarcy.
permeability = 220

# Well spacing. The distance between injector and producer, in meters.
spacing = 800

# Injection temperature. The temperature of the cold water that is being re-injected, in °C.
T_inj = 32

# Flow rate. How much water we extract out of the target layer (and re-inject), in m3/hour.
flowrate = 200

print("\n\nDefinitions complete!")
```

## 3 - Make the doublet
In this cell, we take all the parameters and give them to the doublet function. You can see this happening in the brackets.

After running this  cell, _my_doub_ has all the parameters we defined above.
We will use this in code cell 4.

**Important:** If you change parameters, you have to run this cell again to see the changes.

```python
# Code cell 3 - Pass the details to the function:
my_doub = Doublet(depth_m=depth, thickn_m=thickn, poro_percent=porosity, perm_mD=permeability, w_space=spacing, flow_m3_h=flowrate, T_inj=T_inj)
print('Doublet created')
```

## 4 - Physics
Now we will look at the physics of the doublet. We want to know
1. Production temperature.
2. _Lifetime_ of the doublet. (the number of years for which the production well gives us hot water.)
3. The power that the doublet produces. (We build the doublet because we want power for heating. But the doublet also consumes power, because it costs energy to pump water up and down. The difference between power_in and power_out is how much we actually gain.)

```python
# Production temperature. How hot is the water that we pump from  the target layer to the surface?

Temp = my_doub.T_prod
# (If you like, you can ctrl+click on "T_prd" to see how this calculation is done. It will open doublet.py and jump to the line where this temperature is calculated. This is not necessary for the task.)

print('T_out: %s °C' % round(Temp, 2))
```

```python
# time of thermal breakthrough
lifetime = my_doub.t_breakthrough()
# (If you like, you can ctrl+click on "t_breakthrough()" to see how this calculation is done. It will open doublet.py and jump to the function where the lifetime is calculated. This is not necessary for the task.)

print('Time of thermal breakthrough: %s years' % round(lifetime, 2))
```

```python
# Power

# 1) How much power does the doublet generate?
power_out = my_doub.p_doublet()
# (If you like, you can ctrl+click on "p_doublet()" to see how this calculation is done. It will open doublet.py and jump to the function where the power is calculated. This is not necessary for the task.)
print('Power produced by the doublet: %s MW' % round(power_out, 2))

# 2) how much power is needed to run the pumps?
power_in = my_doub.p_pumps()
print('Power consumed by the doublet: %s MW' % round(power_in, 2))

# 3) The difference:
power_net = power_out - power_in
print('Net power produced: %s MW' % round(power_net, 2))
```
